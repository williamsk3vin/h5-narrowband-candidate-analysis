# Phase 5 — Machine Learning Feasibility Assessment

## Objective

The goal of Phase 5 is to determine whether the candidate-analysis pipeline developed in the previous phases currently provides enough data to justify the use of machine learning.

Instead of applying a machine-learning algorithm simply because candidate features are available, we are evaluating whether the dataset has sufficient sample size, labels, observational diversity, and feature quality to support a scientifically meaningful model.

The machine-learning dataset is defined as:

**One row = one candidate**

Each candidate could be represented using the behavioral and structural features extracted during Phase 4, including:

- Candidate frequency

- Comb spacing

- Frequency drift rate

- Drift-model R²

- Power coefficient of variation

- Power trend

- Power-trend R²

- Persistence

- Number of detection runs

- Longest detection run

- Intermittency

## Current Sample Size

At the current stage of the project, only one candidate has been fully characterized.

Multiple features have been extracted from this candidate, these features describe a single observation rather than a population of independent candidates.

Therefore, the current dataset does not contain enough candidate-level observations to train or  evaluate a machine-learning model.

A large number of features does not compensate for an insufficient number of independent samples.

## Supervised Learning Feasibility

Supervised classification would require trustworthy target labels describing the known class or origin of each candidate.

Possible labels might include categories such as:

- Radio-frequency interference (RFI)

- Instrumental artifacts

- Astrophysical signals

- Other known signal classes

These labels are not currently available.

Without a reliable target variable (`y`), supervised classification is not justified.

Even if labels were obtained in the future, additional issues would need to be considered. These include severe class imbalance, sufficient representation of rare classes, and the reliability of the ground-truth labels.

## Observation Independence and Data Leakage

Future candidate datasets should also contain candidates from many independent observations.

A dataset containing thousands of candidates does not necessarily represent thousands of independent observational conditions. For example, if most candidates originate from a single observation, the apparent dataset size may substantially overstate its actual diversity.

Randomly distributing candidates from the same observation between training and testing sets could also introduce data leakage because those candidates may share telescope conditions, noise characteristics, instrumental effects, and preprocessing artifacts.

Future train/test splits should therefore consider grouping candidates by observation.

## Feature Generalization

Candidate features should also be evaluated for their ability to generalize across observations.

Features such as `candidate_frequency_mhz` may allow a model to learn dataset-specific shortcuts. For example, if known RFI happens to occur within a particular frequency range in the training data, a model could learn that specific frequency range rather than learning the behavioral characteristics of RFI.

Behavioral and structural features may provide more generalizable information. Examples include:

- Persistence

- Intermittency

- Detection run structure

- Comb spacing

- Power variability

- Frequency drift behavior

These describe how a candidate behaves rather than only where it occurs but their generalizability would still need to be tested using candidates collected across many independent observations.





## Unsupervised Learning Feasibility

If a sufficiently large population of candidates becomes available but trustworthy labels remain unavailable, unsupervised machine-learning methods may become appropriate.

Possible approaches include:

- Isolation Forest

- Local Outlier Factor (LOF)

- DBSCAN or other clustering methods

These methods could identify candidates that behave differently from the larger candidate population without requiring predefined target labels.

However, an anomalous candidate should not automatically be interpreted as astrophysical, artificial, or otherwise significant.

An anomaly score indicates only that:

**The candidate is unusual relative to the reference population and the features supplied to the model.**

Possible explanations could still include RFI, instrumental effects, preprocessing artifacts, unusual astrophysical behavior, corrupted data, or other unknown causes.

Therefore, anomaly detection would be most appropriately used as a **candidate-prioritization tool**, rather than as a method for determining the signal's origin.





## Current Decision

Machine learning is not currently justified because only one candidate has been fully characterized.

However, the Phase 4 feature-engineering pipeline establishes a candidate-level representation that could support future machine-learning analysis as additional candidates are processed.

A future workflow could take the form:

Candidate Detection  

→ Candidate Characterization  

→ Feature Extraction  

→ Candidate Feature Table  

→ Unsupervised Anomaly Detection / Clustering  

→ Candidate Prioritization  

→ Further Scientific Investigation

At the current stage, the appropriate decision is to retain the engineered feature pipeline and defer machine-learning modeling until a sufficiently large and diverse candidate population is available.
